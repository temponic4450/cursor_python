```python
# simple
languages = ['python', 'perl', 'c', 'java']

for lang in languages: # for 문
    if lang in ['python', 'perl']: # if 문
        print("%6s need interpreter" % lang) # 문자열 포매팅
    elif lang in ['c', 'java']: # elif 문
        print("%6s need compiler" % lang) # 문자열 포매팅
    else: # else 문
        print("should not reach here")   
```

```python
# coffer (while)
coffee = 10
while True: # boolen 자료형
    money = int(input("돈을 넣어주세요: ")) # 입력받은 값을 int 자료형으로 변환
    if money == 300: # 300원 넣으면 커피 줌
        print("커피를 줍니다.")
        coffee = coffee - 1 # 커피 개수 줄어듦
    elif money > 300: # 300원 초과 시
        print("거스름돈 %d를 주고 커피를 줍니다." % (money - 300)) # 거스름돈 주고 커피 줌
        coffee = coffee - 1 # 커피 개수 줄어듦
    else: # 300원 미만 시
        print("돈을 다시 돌려주고 커피를 주지 않습니다.") # 돈 반환
        print("남은 커피의 양은 %d개입니다." % coffee) # 남은 커피 개수
    if coffee == 0: # 커피 개수 0개 시
        print("커피가 다 떨어졌습니다. 판매를 중지합니다.") # 커피 다 떨어졌다고 알림
        break
```

```python
# continue
a = 0
while a < 10: # 반복문
    a = a + 1
    if a % 2 == 0: continue # 짝수일 경우 반복문 처음으로 돌아감
    print(a)
    
```

```python
# while - else
treeHit = 0
while treeHit < 10: # 반복문
    treeHit = treeHit + 1 # 나무 찍은 횟수 증가
    print("나무를 %d번 찍었습니다." % treeHit) # 나무 찍은 횟수 출력
else:    # 반복문 종료 후 실행
    print("나무 넘어갑니다.") # 나무 넘어감
```

```python
# 중첩 while문
i = 2
while i <= 3: # 반복문
    j = 1
    while j <= 9: # 반복문
        print(f"{i} * {j} = {i*j}") # 구구단 출력 
        j += 1 # j 값 증가
    i += 1 # i 값 증가
```

```python
# for 문
marks = [90, 25, 67, 45, 80] # 리스트

number = 0
for mark in marks: # for 문
    number = number + 1 # number 값 증가
    if mark >= 60: # 60점 이상일 경우
        print(f"{number}번 학생은 합격입니다.") # 합격 출력
    else: # 60점 미만일 경우
        print(f"{number}번 학생은 불합격입니다.") # 불합격 출력
        
```

```python
# for 문
marks = [90, 25, 67, 45, 80] # 리스트
for number in range(len(marks)): # for 문, range 함수 : 0부터 시작, len 함수 : 리스트의 길이
    if marks[number] < 60: # 60점 미만일 경우
        continue # 반복문 처음으로 돌아감
    else: # 60점 이상일 경우
        print(f"{number+1}번 학생 축하합니다. 합격입니다.") # 합격 축하 메시지 출력
        
        
```

```python
# 함수 
def say_myself(name, old, man=True): # 함수 정의
    print(f"나의 이름은 {name}입니다.") # 이름 출력
    print(f"나의 나이는 {old}살입니다.") # 나이 출력
    if man: # 남자일 경우
        print("남자입니다.") # 남자 출력
    else: # 여자일 경우
        print("여자입니다.") # 여자 출력

print("-"*50)
say_myself("박응용", 27) # 함수 호출
print("-"*50)
say_myself("박응용", 27, True) # 함수 호출
```
